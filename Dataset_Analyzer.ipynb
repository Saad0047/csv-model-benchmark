# Dataset Analyzer

## Objective

A terminal-based tool that loads, inspects, cleans, and analyzes a CSV dataset — built to understand how datasets are explored and prepared before Machine Learning.

---

# Milestone 1 — Dataset Loading

- Load a CSV dataset
- Display dataset preview (first 5 rows)
- Show File Name, Number of Rows, Number of Columns, Column Names, Data Types

---

# Milestone 2 — Dataset Inspection

For each column determine:

- Missing Values
- Number of Unique Values
- Data Type

For numerical columns additionally calculate:

- Minimum, Maximum, Mean, Median, Standard Deviation

For categorical columns display:

- Value Counts and Percentage

Automatically classify columns into:

- ID-like (highly unique text columns)
- Categorical ("crayon box")
- Numerical ("pencil box")

---

# Milestone 3 — Data Cleaning

Interactive menu to:

- Remove Duplicate Rows
- Replace Missing Values (mode for categorical columns; mean or median for numerical columns depending on skew)
- Fill columns with a Custom Value

---

# Milestone 4 — Data Analysis

- Highest and Lowest value record per numerical column
- Correlation Matrix (numerical columns)
- Outlier Detection (IQR method)
- Top 10 and Bottom 10 records per numerical column

# DATA-sets we will be choosing: 
Clean-ish, good starting point

Titanic — kaggle.com/competitions/titanic (small, has missing values, mixed types)
Iris — archive.ics.uci.edu/dataset/53/iris (tiny, clean, good baseline test)

Messier, real-world

NYC 311 Service Requests — data.cityofnewyork.us (huge, inconsistent formatting, lots of nulls, date fields)
Airbnb Open Data (Kaggle, search "Airbnb open data") — mixed types, text with commas/quotes, missing values
World Bank Indicators — data.worldbank.org (wide format, tons of NaNs, many years as columns)

Large / stress-test size

NYC Yellow Taxi Trip Data — nyc.gov/site/tlc/about/tlc-trip-record-data.html (millions of rows, good for memory/performance testing)
